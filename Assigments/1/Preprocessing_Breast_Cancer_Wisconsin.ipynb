```python
#imports: 
import numpy as np
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
from sklearn.model_selection import train_test_split
```

```python
#loading dataset:
path = './'
dataset_name = 'breast_cancer.csv'
df = pd.read_csv(path+dataset_name)
df.head()
```

```python
# get insights about the dataset: 
print(f'The shape of the dataset : {df.shape}')
df.info()
```

```python
# let's check null values before the scaling step:
df.isnull().sum().sum() # checking null values 
```

```python
df.isna().sum().sum()# checking nan values
# Prefect! the dataset doesn't have null values
```

```python
#then let's check all the features are numerical values
df.select_dtypes(exclude= 'number').columns# checking what features are not numerical datatype!
#really good, just the label feature is only object data type!
# Then, we are ready to scale the dataset
```

```python
#the dataset without label column
df.iloc[:,: -1]
```

```python
# requirement: scale the dataset between [0,1], so let's use min-max scaler from scikit-learn!
scaler = MinMaxScaler(feature_range=(0,1))
scaled_data = pd.DataFrame(data = scaler.fit_transform(df.iloc[:, : -1]) , columns= df.columns[:-1]) # scaled with converting them back to pandas.
scaled_data.head()
```

```python
#convet label column into 
# malignant --> 1
# benign --> 0
#let's check before and after the number of each class
df['label'].value_counts()
```

```python
#let's practice to do it will apply function
labels_mapping = {'malignant': 1,
 'benign':0}
y = df['label'].apply(lambda x : labels_mapping[x])
y.value_counts() # matched.....  good!!
```

```python
# the step of splitting the data into train, and test sets!!
X_train, X_test , y_train , y_test = train_test_split(scaled_data, y, test_size = 0.25, random_state= 42)
```

```python
# check the sizes:
print(f'The shape of the training samples dataset: {X_train.shape}, and its corresponding labels {y_train.shape}')
print(f'The shape of the test samples dataset: {X_test.shape}, and its corresponding labels {y_test.shape}')
```
