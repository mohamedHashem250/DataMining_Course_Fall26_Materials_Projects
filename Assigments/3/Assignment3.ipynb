```python
# Only using Pandas (20):
# Calculate absolute support for each individual item, such as {Bread}, {Honey}
# we can just count every item by summing the ones in every columns
absolute_support = df.sum()
absolute_support 
```

```python
# Calculate relative support for each individual item
# we can use the aboulate support divided by the number of transactions
relative_support = absolute_support / len(df)
relative_support
```

```python
# Using FP-Growth (25)
# Create a dataframe with frequent itemsets
# Play with the min_sup parameter to see what happens. You will need a lower value to get good outputs.
# Check what is different with Apriori
fpgrowth_df_1 = fpgrowth(df, min_support= 0.5 , null_values= False, use_colnames= True, max_len= None, verbose= 0)  
fpgrowth_df_1 
```

```python
# Using FP-Growth (25)
# Create a dataframe with frequent itemsets
# Play with the min_sup parameter to see what happens. You will need a lower value to get good outputs.
# Check what is different with Apriori
fpgrowth_df_2 = fpgrowth(df, min_support= 0.1 , null_values= False, use_colnames=True, max_len= None, verbose= 0)  
fpgrowth_df_2 
```

```python
# Using association_rules on the above dataframes (30)
# Create association rules using the Apriori frequent itemset dataframe (minimum threshold of 0.01)
# Create association rules using the FP-Growth frequent itemset dataframe (minimum threshold of 0.01)

association_rules_apriori= association_rules(apriori_df_2 , 
    num_itemsets= len(df), 
    metric="confidence",   # You can also use "lift" or "support" based on your needs
    min_threshold=0.01, 
    null_values= False, 
    support_only= False, 
    )

association_rules_apriori
```

```python
association_rules_apriori
```

```python
#Create association rules using the FP-Growth frequent itemset dataframe (minimum threshold of 0.01)
association_rules_fpgrowth= association_rules(fpgrowth_df_2  , 
    num_itemsets= len(df), 
    metric="confidence",   # You can also use "lift" or "support" based on your needs
    min_threshold=0.01, 
    null_values= False, 
    support_only= False, 
)
association_rules_fpgrowth
```

```python
for _, row in association_rules_apriori.iterrows():
    print(f"{row['antecedents']} => {row['consequents']} "
          f"[{row['support']:.2f}, {row['confidence']:.2f}, {row['lift']:.2f}]")
```

```python
def imbalance_ratio(support_a, support_b, support_ab):
    numerator = abs(support_a - support_b)
    denominator = support_a + support_b - support_ab

    return numerator / denominator
```

```python
association_rules_apriori['imbalance ratio'] = association_rules_apriori.apply(
    lambda row: imbalance_ratio(
        row['antecedent support'],
        row['consequent support'],
        row['support']
    ),
    axis=1
)

association_rules_apriori.head()
```

```python
association_rules_fpgrowth['imbalance ratio'] = association_rules_fpgrowth.apply(
    lambda row: imbalance_ratio(
        row['antecedent support'],
        row['consequent support'],
        row['support']
    ),
    axis=1
)

association_rules_fpgrowth.head()
```
